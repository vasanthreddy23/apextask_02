```python

# Step 1: Upload your cleaned Excel file
from google.colab import files
uploaded = files.upload()   # choose cleaned_sales_vas (1).xlsx

# Step 2: Import libraries
import pandas as pd
import sqlite3

# Step 3: Load Excel file (use exact uploaded name)
df = pd.read_excel("cleaned_sales_vas (1).xlsx")

# Step 4: Create SQLite database connection
conn = sqlite3.connect("sales_vas.db")

# Step 5: Save DataFrame to SQLite table
df.to_sql("sales_vas", conn, if_exists="replace", index=False)

# -------------------------------
# SQL FUNDAMENTALS (Day 7-8)
# -------------------------------

# SELECT + WHERE
query1 = "SELECT * FROM sales_vas WHERE Sales > 500;"
print(pd.read_sql(query1, conn))

# ORDER BY + LIMIT
query2 = "SELECT Name, Item, Sales FROM sales_vas ORDER BY Sales DESC LIMIT 5;"
print(pd.read_sql(query2, conn))

# GROUP BY + HAVING
query3 = """
SELECT Item, SUM(Sales) AS TotalSales
FROM sales_vas
GROUP BY Item
HAVING SUM(Sales) > 1000
ORDER BY TotalSales DESC;
"""
print(pd.read_sql(query3, conn))

# JOIN Example (self-join for demo)
query4 = """
SELECT a.Name, a.Item, b.Item AS OtherItem
FROM sales_vas a
JOIN sales_vas b ON a.Name = b.Name
WHERE a.Item <> b.Item;
"""
print(pd.read_sql(query4, conn))

# -------------------------------
# ADVANCED SQL (Day 9-10)
# -------------------------------

# Subquery
query5 = """
SELECT Name, Sales
FROM sales_vas
WHERE Sales > (SELECT AVG(Sales) FROM sales_vas);
"""
print(pd.read_sql(query5, conn))

# CTE (WITH clause)
query6 = """
WITH MonthlySales AS (
    SELECT Date, SUM(Sales) AS TotalSales
    FROM sales_vas
    GROUP BY Date
)
SELECT * FROM MonthlySales ORDER BY Date;
"""
print(pd.read_sql(query6, conn))

# Window Functions
query7 = """
SELECT Name, Item, Sales,
       RANK() OVER (ORDER BY Sales DESC) AS SalesRank,
       ROW_NUMBER() OVER (ORDER BY Profit DESC) AS ProfitRow
FROM sales_vas;
"""
print(pd.read_sql(query7, conn))

# Views (create reusable query)
conn.execute("""
CREATE VIEW IF NOT EXISTS TopCustomers AS
SELECT Name, SUM(Sales) AS TotalSpend
FROM sales_vas
GROUP BY Name
ORDER BY TotalSpend DESC;
""")
print(pd.read_sql("SELECT * FROM TopCustomers;", conn))

# -------------------------------
# BUSINESS QUESTIONS (Day 11-13)
# -------------------------------

# Top 5 products by sales
q1 = """
SELECT Item, SUM(Sales) AS TotalSales
FROM sales_vas
GROUP BY Item
ORDER BY TotalSales DESC
LIMIT 5;
"""
print("Top 5 Products by Sales:\n", pd.read_sql(q1, conn))

# Monthly sales trend
q2 = """
SELECT Date, SUM(Sales) AS TotalSales
FROM sales_vas
GROUP BY Date
ORDER BY Date;
"""
print("\nMonthly Sales Trend:\n", pd.read_sql(q2, conn))

# Customer segmentation by spend
q3 = """
SELECT Name, SUM(Sales) AS TotalSpend
FROM sales_vas
GROUP BY Name
ORDER BY TotalSpend DESC;
"""
print("\nCustomer Segmentation:\n", pd.read_sql(q3, conn))

# -------------------------------
# Close connection
# -------------------------------
conn.close()
```
